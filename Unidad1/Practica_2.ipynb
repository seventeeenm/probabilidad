```python
#importar librerias
import random
import matplotlib.pyplot as plt
from statistics import *
```

```python
#Generamos 100 numeros aleatorios que representan la produccion en metros de  tela de 100 dias
fabrica1 =[]
for i in range(100):
  numero_aleatorio = round(random.uniform(50,100),2)
  fabrica1.append(numero_aleatorio)
  print(fabrica1)
```

```python
#Generamos 100 numeros aleatorios que representan la produccion en metros de  tela de 100 dias
fabrica2 =[]
for i in range(100):
  numero_aleatorio = round(random.uniform(50,100),2)
  fabrica2.append(numero_aleatorio)
  print(fabrica2)
```

**Maximo, minimo y rango**

```python
#Fabrica 1
maximo = max(fabrica1)
minimo = min(fabrica1)
rango = maximo - minimo
mediana_fabrica1 = mean(fabrica1)
media_fabrica1=mean(fabrica1)
moda_fabrica1=multimode(fabrica1)
desviacion_estandar_fabrica1=stdev(fabrica1)
print("---Estadistica descriptiva de la Fábrica 1---")
print("Máximo:", maximo)
print("Mínimo:", minimo)
print("Rango:", rango)
print("Media:", media_fabrica1)
print("Mediana:", mediana_fabrica1)
print("Moda:", moda_fabrica1)
print("La desviacion estandar de la fabrica 1 es:",desviacion_estandar_fabrica1)

#Fabrica 2
maximo2 = max(fabrica2)
minimo2 = min(fabrica2)
rango2 = maximo2 - minimo2
mediana_fabrica2 = mean(fabrica2)
media_fabrica2=mean(fabrica2)
moda_fabrica2=multimode(fabrica2)
desviacion_estandar_fabrica2=stdev(fabrica2)
print("---Estadistica descriptiva de la Fábrica 2---")
print("Máximo:", maximo2)
print("Mínimo:", minimo2)
print("Rango:", rango2)
print("Media:", media_fabrica2)
print("Mediana:", mediana_fabrica2)
print("Moda:", moda_fabrica2)
print("La desviacion estandar de la fabrica 2 es:",desviacion_estandar_fabrica2)
```

**Graficamos**

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# --- Histograma Fábrica 1 ---
ax1.hist(fabrica1, bins=10, color="skyblue", edgecolor="black")
ax1.set_title("Produccion diaria de la fabrica 1")
ax1.set_xlabel("Valor")
ax1.set_ylabel("Frecuencia")

# Líneas estadísticas para Fábrica 1
ax1.axvline(
    media_fabrica1, color="red", linestyle="dashed", linewidth=2, label=f"Media: {media_fabrica1:.2f}"
)
ax1.axvline(
    mediana_fabrica1,
    color="green",
    linestyle="dotted",
    linewidth=2,
    label=f"Mediana: {mediana_fabrica1:.2f}",
)
ax1.axvline(
    moda_fabrica1[0],
    color="purple",
    linestyle="dashdot",
    linewidth=2,
    label=f"Moda: {moda_fabrica1[0]}", #Aqui se usa el 0 para mostrar la primera moda pq hay multiples.
)

ax1.axvline(
    media_fabrica1 + desviacion_estandar_fabrica1,
    color="black",
    linestyle="solid",
    linewidth=1.5,
    label=f"Desv. Estándar (+): {desviacion_estandar_fabrica1:.2f}",
)
ax1.axvline(
    media_fabrica1 - desviacion_estandar_fabrica1,
    color="black",
    linestyle="solid",
    linewidth=1.5,
)

ax1.legend(loc="upper right", fontsize=9)


# --- Histograma Fábrica 2 ---
ax2.hist(fabrica2, bins=10, color="salmon", edgecolor="black")
ax2.set_title("Produccion diaria de la fabrica 2")
ax2.set_xlabel("Valor")
ax2.set_ylabel("Frecuencia")

# Líneas estadísticas para Fábrica 2
ax2.axvline(
    media_fabrica2, color="red", linestyle="dashed", linewidth=2, label=f"Media: {media_fabrica2:.2f}"
)
ax2.axvline(
    mediana_fabrica2,
    color="green",
    linestyle="dotted",
    linewidth=2,
    label=f"Mediana: {mediana_fabrica2:.2f}",
)
ax2.axvline(
    moda_fabrica2[0],
    color="purple",
    linestyle="dashdot",
    linewidth=2,
    label=f"Moda: {moda_fabrica2[0]}",
)

ax2.axvline(
    media_fabrica2 + desviacion_estandar_fabrica2,
    color="black",
    linestyle="solid",
    linewidth=1.5,
    label=f"Desv. Estándar (+): {desviacion_estandar_fabrica2:.2f}",
)
ax2.axvline(
    media_fabrica2 - desviacion_estandar_fabrica2,
    color="black",
    linestyle="solid",
    linewidth=1.5,
)

ax2.legend(loc="upper right", fontsize=9)

plt.tight_layout()
plt.show()
```

**Diagrama de densidad**

```python
#La densidad grafica es una forma de visualizar la distribucion de datos
#un intervalo o periodo de tiempo continuo.
#Se basa en un histograma que usa el suavizado de kernel para trazar valores,
#lo que permite obtener distribuciones mas suaves al reducir el ruido.
#Una curva de densidad muestra la ubicacion de la media y la mediana de una distribucion.
import seaborn as sns
sns.histplot(x=fabrica1, kde=True, kde_kws={"bw_adjust": 0.5}, bins=10)
```
