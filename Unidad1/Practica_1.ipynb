**Ejemplo 1: Registro de calificaciones de 12 alumnos**
1. en este ejemplo generamos una lista que contiene las calificaciones de 12 alumnos.
2. con esa lista de valores se pretende determinar las medidas de tendencia centarl, es decir, la media,mediana y moda.

```python
#Importar libreria
from statistics import *
#Lista de calificaciones del grupo "a" de probabilidad y estadistica
grupoa = [70,70,65,90,100,95,45,80,73,20,100,96]
print(grupoa)
#la funcion type nos devuelve el tipo de dato de la variable
type(grupoa)
#Sacar la media de las calificaciones con una formula
promedio=sum(grupoa)/len(grupoa)
print(promedio)
```

```python
#Utilizar la funcion mean que esta en la libreria statistics para sacar el promedio
promedio2 =mean(grupoa)
print(promedio2)
```

###Determinar la mediana
La media, tambien conocida como promedio, es el valor que se obtiene al dividir la suma de un conjunto de numeros entre la cantidad de ellos

```python
#Determinar el valor de la mediana con la funcion median
mediana=median(grupoa)
print(mediana)
```

###Determinar la moda
1. La moda es el valor que mas se repite dentro de un conjunto de datos
2. A diferencia de la media y la mediana, la moda no requiere valores numericos y puede utilizarse en datos categoricos e discretos.

```python
#Determinar el valor de la moda con la funcion mode, en caso de que hayan dos modas se usan la funcion multimode
moda=multimode(grupoa)
print(grupoa)
print("la moda son:")
print(moda)
```

**Ejemplo 2. Tiempo de 100 metros de un grupo de 12 alumnos**
1. En este ejemplo generamos una lista de un grupo de 10 alumnos que corren los 100 metros y se registran los tiempos de cada uno.
2. Con esta lista de valores se pretende determinar las medidas de tendencia centra, es decir, la media,mediana y moda

**Ejemplo 3: Histograma de 2 grupos**
1. En este ejemplo generamos 100 numeros aleatorios reales que representa la produccion diaria de 2 fabricas de telas (medidas en metros) durante 100 dias.
2. Generar otra lista con la produccion de 100 telares para una segunda fabrica.
3. Con estas listas de valores se pretende determinar las medidas de tendencia centar, es decir, media, moda, mediana y la desviacion estandar.
4. Graficar un histograma para cada fabrica.

```python
#Importamos libreria para generar numeros aleatorios y para graficar
import random
import matplotlib.pyplot as plt

#Generamos los datos para la fabrica 1
fabrica1 = []
for i in range(100):
    numero_aleatorio = round(random.uniform(50, 100), 2)
    fabrica1.append(numero_aleatorio)

#Generamos los datos para la fabrica 2
fabrica2 = []
for i in range(100):
    numero_aleatorio = round(random.uniform(70, 100), 2)
    fabrica2.append(numero_aleatorio)

#Gráfica de la Fábrica 1
plt.figure(figsize=(8, 4))
plt.hist(fabrica1, bins=10, color="skyblue", edgecolor="black")
plt.title("Produccion diaria de la fabrica 1")
plt.xlabel("Metros producidos")
plt.ylabel("Frecuencia")
plt.show()

#Gráfica de la Fábrica 2
plt.figure(figsize=(8, 4))
plt.hist(fabrica2, bins=10, color="salmon", edgecolor="black")
plt.title("Produccion diaria de la fabrica 2")
plt.xlabel("Metros producidos")
plt.ylabel("Frecuencia")
plt.show()
```

```python
#Calculamos la media, mediana y la moda de ambas fabricas
media1 = mean(fabrica1)
media2 = mean(fabrica2)
mediana1 = median(fabrica1)
mediana2 = median(fabrica2)
moda1 = multimode(fabrica1)
moda2 = multimode(fabrica2)

print("Valores de la frabica 1")
print("Media:", media1)
print("Mediana:", mediana1)
print("Moda:", moda1)

print("Valores de la frabica 2")
print("Media:", media2)
print("Mediana:", mediana2)
print("Moda:", moda2)
```

```python
#Generar 2 subplot de 1 renglon por 2 columnas para despeglar los 2 histogramas.
#una a lado de la otra
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2)
```

```python
#Generar 2 subplot de 1 renglon por 2 columnas para despeglar los 2 histogramas.
#una a lado de la otra
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10, 5))

#Histograma Fábrica 1
ax1.hist(fabrica1, bins=10, color="skyblue", edgecolor="black")
ax1.set_title("Produccion diaria de la fabrica 1")
ax1.set_xlabel("Metros producidos")
ax1.set_ylabel("Frecuencia")

#Histograma Fábrica 2
ax2.hist(fabrica2, bins=10, color="salmon", edgecolor="black")
ax2.set_title("Produccion diaria de la fabrica 2")
ax2.set_xlabel("Metros producidos")
ax2.set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

```python
# Generar 2 subplot de 1 renglón por 2 columnas con media, mediana y moda
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# --- Histograma Fábrica 1 ---
ax1.hist(fabrica1, bins=10, color="skyblue", edgecolor="black")
ax1.set_title("Produccion diaria de la fabrica 1")
ax1.set_xlabel("Valor")
ax1.set_ylabel("Frecuencia")

#media y mediana para Fábrica 1
ax1.axvline(
    media1, color="red", linestyle="dashed", linewidth=2, label=f"Media: {media1:.2f}"
)
ax1.axvline(
    mediana1,
    color="green",
    linestyle="dotted",
    linewidth=2,
    label=f"Mediana: {mediana1:.2f}",
)
ax1.legend()
#moda fabrica 1
ax1.axvline(
    moda1[0],
    color="purple",
    linestyle="dashdot",
    linewidth=2,
    label=f"Moda: {moda1}",
)
ax1.legend()


# --- Histograma Fábrica 2 ---
ax2.hist(fabrica2, bins=10, color="salmon", edgecolor="black")
ax2.set_title("Produccion diaria de la fabrica 2")
ax2.set_xlabel("Valor")
ax2.set_ylabel("Frecuencia")

#media y mediana para Fábrica 2
ax2.axvline(
    media2, color="red", linestyle="dashed", linewidth=2, label=f"Media: {media2:.2f}"
)
ax2.axvline(
    mediana2,
    color="green",
    linestyle="dotted",
    linewidth=2,
    label=f"Mediana: {mediana2:.2f}",
)
ax2.legend()

#moda fabrica 2
ax2.axvline(
    moda2[0],
    color="purple",
    linestyle="dashdot",
    linewidth=2,
    label=f"Moda: {moda2}",
)
ax2.legend()


plt.tight_layout()
plt.show()
```
