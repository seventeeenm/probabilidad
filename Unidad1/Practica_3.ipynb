```python
#Basado en el libro estadistica para administracion pagina 12
#Creamos una lista vacia que representa la edad de 100 personas.
import random
edades= []
#Hacemos un ciclo para generar 100 numeros aleatorios que representen la edad de las personas
for i in range(100):
  #Generamos un numero aleatorio con la funcion randint y lo guardamos en la variable numero_aleatorio
  numero_aleatorio = random.randint(18,70)
  #Agregamos el numero aleatorio a la lista de valores
  edades.append(numero_aleatorio)
#Imprimimos la lista de edades
print(edades)
```

```python
import pandas as pd
#Crear un DataFrame de pandas con los numeros aleatorios
df = pd.DataFrame(edades, columns=['Edades'])
df.head()
```

```python
#Calcular la tabla de distribucion de frecuencias
tabla_frecuencias = df['Edades'].value_counts().reset_index()
tabla_frecuencias.columns = ['Edades', 'Frecuencia']
tabla_frecuencias = tabla_frecuencias.sort_values(by='Edades')
#Imprimir la tabla de distribucion de frecuencias
print("Tabla de distribucion de frecuencias:")
tabla_frecuencias
```

<mark>**Calcular la frecuencia relativa**</mark>

```python
tabla_frecuencias['Frecuencia_relativa'] = tabla_frecuencias['Frecuencia'] / 100
tabla_frecuencias
```

<mark>**Calcular la frecuencia porcentual**</mark>

```python
tabla_frecuencias['Frecuencia_porcentual'] = tabla_frecuencias['Frecuencia'] * 100
tabla_frecuencias
```

<mark>**Calcular la frecuencia acumulada**</mark>

```python
tabla_frecuencias['Frecuencia_acumulada'] = tabla_frecuencias['Frecuencia'].cumsum()
tabla_frecuencias
```

Regla de sturges
1. Es una regla que sirve para calcular el numero de clases o intervalos idoneo en los que se debe dividir un conjunto de datos.
2. Establece que **el numero de clases es igual a uno mas que el logaritmo en base dos del numero total de datos**

c=1+log2(N)
donde N es el tamño de la muestra

```python
#Importamos la libreria math y volvemos a la usar la lista de edades
import math
print(edades)
```

```python
#Creamos un DataFrame con las edades
df=pd.DataFrame({'Edad0': edades})
df
```

```python
#Calcular el numero de intervalos utilizando la regla de sturges
n= len(edades)
K= int(1+math.log2(n))
print("Valor de N", K)
print("El numero de intervalos es:", K)
```

```python
# Crear la tabla de frecuencia
tabla_frecuencia = pd.cut(df['Edad0'], bins=K, include_lowest=True).value_counts().reset_index()
tabla_frecuencia.columns = ['Clase', 'Frec']
tabla_frecuencia
```

```python
#Ordenar la tabla de frecuencia por el intervalo
tabla_frecuencia = tabla_frecuencia.sort_values(by='Clase')
tabla_frecuencia
```

```python
#Genrar columnas siguientes
tabla_frecuencia['Frec Relativa']=tabla_frecuencia['Frec']/n
tabla_frecuencia['Frec Acumulada']=tabla_frecuencia['Frec'].cumsum()
tabla_frecuencia['Freciencia porcentual']=tabla_frecuencia['Frec']/100
tabla_frecuencia['Frecuencia relativa acumulada'] = tabla_frecuencia['Frec Relativa'].cumsum()
tabla_frecuencia['Frecuencia porcentual acumulada'] = tabla_frecuencia['Freciencia porcentual'].cumsum()
#Mostrar la tabla de frecuencia
print(tabla_frecuencia)
```
