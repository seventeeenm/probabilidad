```python
import numpy as np
import matplotlib.pyplot as plt

# datos
X = np.array([0, 10, 20, 50])
P_x = np.array([0.50, 0.30, 0.15, 0.05])

# 1. Checar que las probabilidades sumen 1
suma_prob = np.sum(P_x)
print(f"1. La suma de las probabilidades da: {suma_prob:.2f}")

# 2. Gráfica de la distribución de probabilidad
plt.figure(figsize=(8, 4))
plt.bar(X, P_x, color='#4A7BC7', edgecolor='black', width=4)
plt.title("2. Distribución de probabilidad de los Premios", fontsize=14)
plt.xlabel("Monto del premio en pesos (X)")
plt.ylabel("Probabilidad P(X)")
plt.xticks(X)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 3. Sacar el valor esperado, varianza y desviación
E_x = np.sum(X * P_x)
varianza = np.sum(((X - E_x)**2) * P_x)
desviacion = np.sqrt(varianza)

print(f"\n3. Resultados de los cálculos:")
print(f"   Valor esperado (Ganancia promedio): ${E_x:.2f}")
print(f"   Varianza: {varianza:.4f}")
print(f"   Desviación estándar: ${desviacion:.4f}")

# 4. Función acumulada F(x) y su gráfica escalonada
F_x = np.cumsum(P_x)
print(f"\n4. Función de distribución acumulada F(x):")
for i in range(len(X)):
    print(f"   F({X[i]}) = {F_x[i]:.2f}")

plt.figure(figsize=(8, 4))
plt.step(X, F_x, where='post', color='#E05A47', linewidth=2)
plt.plot(X, F_x, 'o', color='#E05A47', alpha=0.8)
plt.title("4. Gráfica de la función acumulada F(x)", fontsize=14)
plt.xlabel("Monto del premio en pesos (x)")
plt.ylabel("Acumulada F(x)")
plt.xticks(X)
plt.yticks(np.arange(0, 1.1, 0.1))
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# 5. Interpretación
print("\n5. ¿Qué significan estos resultados?")
print(f"   • ¿Qué significa el valor esperado? Salió ${E_x:.2f}. Esto significa que, si jugara muchísimas veces en esta tómbola, en promedio ganaría unos $8.50 pesos por cada boleto que saque.")
print(f"   • ¿El juego es favorable para el jugador? Depende de cuánto cueste el boleto. Como en promedio ganaria $8.50, si me cobran por ejemplo $10 o $15 pesos por jugar, el juego no es favorable. Si el boleto es gratis o cuesta menos de $8.50, entonces sí.")
```
