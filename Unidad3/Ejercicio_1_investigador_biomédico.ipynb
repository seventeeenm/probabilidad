```python
import numpy as np
import matplotlib.pyplot as plt

# Datos
X = np.array([0, 1, 2, 3, 4, 5])
P_x = np.array([0.10, 0.25, 0.30, 0.20, 0.10, 0.05])

# 1. Checar que las probabilidades sumen 1
suma_prob = np.sum(P_x)
print(f"1. La suma de las probabilidades da: {suma_prob:.2f}")

# 2. Gráfica de la distribución de probabilidad
plt.figure(figsize=(8, 4))
plt.bar(X, P_x, color='#4A7BC7', edgecolor='black')
plt.title("2. Distribución de probabilidad", fontsize=14)
plt.xlabel("Células anormales (X)")
plt.ylabel("Probabilidad P(X)")
plt.xticks(X)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 3. Sacar el valor esperado, varianza y desviación
E_x = np.sum(X * P_x)
varianza = np.sum(((X - E_x)**2) * P_x)
desviacion = np.sqrt(varianza)

print(f"\n3. Resultados de los cálculos:")
print(f"   Valor esperado (Promedio): {E_x:.2f}")
print(f"   Varianza: {varianza:.4f}")
print(f"   Desviación estándar: {desviacion:.4f}")

# 4. Función acumulada F(x) y su gráfica escalonada
F_x = np.cumsum(P_x)
print(f"\n4. Función de distribución acumulada F(x):")
for i in range(len(X)):
    print(f"   F({X[i]}) = {F_x[i]:.2f}")

plt.figure(figsize=(8, 4))
plt.step(X, F_x, where='post', color='#E05A47', linewidth=2)
plt.plot(X, F_x, 'o', color='#E05A47', alpha=0.8)
plt.title("4. Gráfica de la función acumulada F(x)", fontsize=14)
plt.xlabel("Células anormales (x)")
plt.ylabel("Acumulada F(x)")
plt.xticks(X)
plt.yticks(np.arange(0, 1.1, 0.1))
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# 5. Interpretación
print("\n5. ¿Qué significan estos resultados?")
print(f"   • Promedio: Salió {E_x:.2f}. O sea, lo normal sería encontrar unas 2 células anormales por muestra de tejido.")
print(f"   • Variabilidad: La desviación es de {desviacion:.2f}. No hay tanta variación; la gran mayoría de las muestras van a tener entre 1 y 3 células anormales.")
print(f"   • Función acumulada: Básicamente nos dice que es más común ver poquitas células anormales. Por ejemplo, hay un 65% de chance de toparse con 2 o menos, y encontrar 5 es algo bastante raro (solo el 5% de las veces).")
```
