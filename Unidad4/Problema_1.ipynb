```python
import statistics

# Datos
media = 70
desviacion = 4
total_estudiantes = 500

# Distribución normal
distribucion = statistics.NormalDist(mu=media, sigma=desviacion)

# 1. Probabilidad entre 68 y 72 kg
prob_superior = distribucion.cdf(72)
prob_inferior = distribucion.cdf(68)
probabilidad = prob_superior - prob_inferior

# 2. Número de estudiantes
estudiantes = probabilidad * total_estudiantes

print(f"Probabilidad de estudiantes que pesan más de 68 y menos de 72 kg: {probabilidad:.4f} ({probabilidad*100:.2f}%)")
print(f"Número de estudiantes: aproximadamente {round(estudiantes)} estudiantes")
```
