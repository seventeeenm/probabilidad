##Que tipo de distribucion es?
Es de tipo poisson, porque modela la ocurrencia de eventos discretos (ingreso de pacientes) en un intervalo continuo de tiempo (un día), con una tasa de ocurrencia promedio conocida ($\lambda = 5$).

```python
from scipy.stats import poisson

# Datos
media_lambda = 5

# a) Probabilidad de que ingresen exactamente 2 personas: P(X = 2)
probabilidad_2 = poisson.pmf(2, media_lambda)

# b) Probabilidad de que x exceda de 10: P(X > 10) = 1 - P(X <= 10)
probabilidad_mayor_10 = 1 - poisson.cdf(10, media_lambda)

#resultados
print(f"a) La probabilidad de que ingresen exactamente 2 personas es: {probabilidad_2:.4f} (o {probabilidad_2 * 100:.2f}%)")
print(f"b) La probabilidad de que x exceda de 10 personas es: {probabilidad_mayor_10:.4f} (o {probabilidad_mayor_10 * 100:.2f}%)")

# Explicacion
print("No es probable que el número de ingresos exceda de 10 personas, ya que la probabilidad calculada es muy baja (1.37%).")
```
