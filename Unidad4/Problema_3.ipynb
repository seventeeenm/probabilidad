##Que tipo de distribucion es?
Es tipo binomial, porque existe un número fijo y conocido de ensayos o tamaño de muestra (n = 1000$ motores en el embarque), cada ensayo tiene solo dos resultados posibles: el motor es defectuoso o no lo es y existe una probabilidad constante de que ocurra el evento de interés (p = 0.001 de que resulte no satisfactorio).

```python
from scipy.stats import binom

# Datos
n_motores = 1000        # Número de ensayos
p_defectuoso = 0.001    # Probabilidad de que un motor sea defectuoso

# a) Probabilidad de que ninguno sea defectuoso
prob_0 = binom.pmf(0, n_motores, p_defectuoso)

# b) Probabilidad de que tres sean defectuosos
prob_3 = binom.pmf(3, n_motores, p_defectuoso)

# c) Probabilidad de que cuatro sean defectuosos
prob_4 = binom.pmf(4, n_motores, p_defectuoso)

#resultados
print(f"Probabilidad de que ninguno sea defectuoso (0): {prob_0:.4f} (o {prob_0 * 100:.2f}%)")
print(f"Probabilidad de que tres sean defectuosos (3): {prob_3:.4f} (o {prob_3 * 100:.2f}%)")
print(f"Probabilidad de que cuatro sean defectuosos (4): {prob_4:.4f} (o {prob_4 * 100:.2f}%)")
```
